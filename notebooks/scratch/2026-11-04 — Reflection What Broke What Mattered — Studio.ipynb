### 1. Data quality

Which single data-quality problem, if it had gone unnoticed, would have changed your conclusion the most? Give the number both ways.

One data quality issue we encountered was fragmented SKUs, where the same products appeared under multiple different SKUs. To address this issue, we consolidated SKUs representing the same products into one standardized SKU. Before consolidation, we had 45 distinct SKUs, which were reduced to 20 canonical products after cleaning. From our Q2 analysis, we found that four different SKUs (POPTART-STRAW, PT-12, PT_STRAWBERRY, and SKU#459812) represented the same Strawberry Pop-Tarts product, which was the largest consolidated SKU family. Before consolidation, these four SKUs recorded 1,239, 1,332, 1,230, and 1,204 units sold, respectively, based on transactions with valid dates. After consolidation, they were combined into a single standardized SKU, POPTART-STRAW, showing a total of 5,005 units sold. This allowed us to accurately measure the total sales of Strawberry Pop-Tarts as one product rather than treating the four SKUs as separate products. If SKU fragmentation had gone unoticed, it could have changed our conclusions the most because we might have underestimated overall product demand by analyzing the same product under multiple separate products. For example, the largest individual Strawberry Pop-Tarts SKU recorded only 1,332 units sold, compared to 5,005 units across all four SKUs after consolidation. This could have led us to incorrectly identify which products had the highest daily sales and which experienced the greatest increases in average daily sales from the baseline to the surge period.

### 2. ETL trade-offs

Name a cleaning decision where a reasonable person could have chosen the other way. What was the other option, what would it have done to your reported numbers, and would you make the same call again?

After converting timestamps to datetime and missing or invalid timestamps to NaT, we decided to keep rows with missing timestamps during the initial data cleaning because they still contained useful transaction information, such as SKU, product name, quantity, and unit price. We wanted to preserve this information because removing these rows could lead to an underestimation of overall sales and revenue in analyses that did not require dates.
The other option was to remove rows with missing timestamps during initial cleaning, which would have reduced the number of available transactions. However, our analytical questions primarily focused on daily sales trends and comparisons between the baseline and surge periods, which required valid transaction dates. After merging with the weather API data, we excluded 881 rows with missing dates for the weather-based analysis and other date-dependent analyses. In Q2, we excluded transactions with missing dates from both the fragmented and consolidated Strawberry Pop-Tarts sales comparisons to ensure consistency with our daily sales analysis. Therefore, removing these rows earlier would not have changed our reported numbers. We would make the same call again to preserve useful information for other potential analyses.
